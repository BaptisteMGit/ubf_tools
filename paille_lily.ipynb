```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
rho_air = 1.293
rho_eau = 1000
g = 9.81
H_paille = 20 * 1e-2
r_paille = 5 * 1e-3
S =  2 * np.pi * r_paille**2
P0 = 1.013 * 1e5 
z0 = 10 * 1e-2
```

```python
a = -rho_eau * g
b = H_paille*rho_eau*g + P0
c = -P0*z0

delta = b**2 - 4*a*c
# print(delta)
z_eq1= (-b - np.sqrt(delta)) / (2 * a)
z_eq2 = (-b + np.sqrt(delta)) / (2 * a)

print(z_eq1, z_eq2)
print(z_eq2 * 1e2)
z_eq = z_eq2
delta_z = z0-z_eq
print(f"Variation de z jusqu'a l'équilibre : {delta_z*1e3} mm")
```

```python
H_min = 1 * 1e-2
H_max = 100
H = np.linspace(H_min, H_max, 10000)
zinit = 1 / 2 * H 
zinit = 2 / 3 * H 

a = -rho_eau * g
b = H * rho_eau * g + P0
c = -P0 * zinit

delta = b**2 - 4 * a * c
# print(delta)
z_eq1 = (-b - np.sqrt(delta)) / (2 * a)
z_eq2 = (-b + np.sqrt(delta)) / (2 * a)

plt.figure()
# plt.plot(H, z_eq1, label="z_eq1")
plt.plot(H, z_eq2, label="z_eq2")
plt.xlabel("Hauteur paille [m]")
```

```python
delta_z = zinit - z_eq2
delta_z_rel = delta_z / zinit * 100

plt.figure()
plt.plot(H, delta_z_rel)
plt.xlabel("Hauteur paille [m]")
plt.ylabel("Variation relative de la hauteur d'eau dans la paille [%]")
```
